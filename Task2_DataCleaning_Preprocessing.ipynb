# Missing Value Handling



```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.drop("Cabin", axis=1, inplace=True)
```

```python
df.isnull().sum()
```

### Observation

- Age missing values filled using Median.
- Embarked missing values filled using Mode.
- Cabin column dropped due to excessive missing values.
- Dataset now contains no missing values.

# Duplicate Record Check

```python
duplicate_count = df.duplicated().sum()

print("Number of Duplicate Records:", duplicate_count)
```

```python
df = df.drop_duplicates()

print("Duplicates After Removal:", df.duplicated().sum())
```

# Data Type Verification

```python
df.dtypes
```

# Outlier Detection

```python
Q1 = df["Fare"].quantile(0.25)

Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("Number of Outliers:", len(outliers))
```

```python
df = df[
    (df["Fare"] >= lower_bound) &
    (df["Fare"] <= upper_bound)
]
```

```python
print("Shape Before Outlier Removal:", df.shape)
```

### Observation

- IQR method was used for outlier detection.
- 116 Fare outliers were identified.
- Dataset size reduced from 891 rows to 775 rows after outlier treatment.

# Feature Scaling


```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()

columns_to_scale = [
    "Age",
    "Fare",
    "SibSp",
    "Parch"
]

df[columns_to_scale] = scaler.fit_transform(
    df[columns_to_scale]
)

df.head()
```

# Export Clean Dataset
